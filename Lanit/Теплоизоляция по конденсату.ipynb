## В.2.4 Расчет толщины изоляции, предотвращающей конденсацию влаги из воздуха на ее поверхности

Данный расчет производится для изолированных объектов, расположенных в помещениях и содержащих вещества с температурой ниже температуры окружающего воздуха.

В этом случае изоляция должна обеспечивать требуемый расчетный перепад между температурами наружного воздуха и поверхностью изоляции $(t_н - t_п)$, при котором исключается конденсация влаги из воздуха (таблица В.4).

### Таблица В.4 — Расчетный перепад $t_н - t_п$, °C

| $t_н$, °C | $\varphi = 40\%$ | $\varphi = 50\%$ | $\varphi = 60\%$ | $\varphi = 70\%$ | $\varphi = 80\%$ | $\varphi = 90\%$ |
|---:|---:|---:|---:|---:|---:|---:|
| 10 | 13,4 | 10,4 | 7,8 | 5,5 | 3,5 | 1,6 |
| 15 | 14,2 | 10,9 | 9,1 | 5,7 | 3,6 | 1,7 |

Расчет выполняется по формулам:

### Для плоской поверхности

$$
R_{из}
=
\frac{t_п-t_в}{t_н-t_п}R_н
$$

$$
\delta_{из}
=
\frac{\lambda_{из}(t_н-t_в)}
{\alpha_н(t_н-t_п)}
$$

где:

- $R_{из}$ — термическое сопротивление слоя изоляции;
- $R_н$ — термическое сопротивление теплоотдаче с наружной поверхности;
- $\lambda_{из}$ — коэффициент теплопроводности изоляции;
- $\alpha_н$ — коэффициент теплоотдачи с наружной поверхности;
- $t_н$ — температура наружного воздуха;
- $t_п$ — температура поверхности изоляции;
- $t_в$ — температура вещества внутри объекта.

### Для цилиндрической поверхности

$$
R_{из}^{L}
=
\frac{t_п-t_в}{t_н-t_п}R_н^{L}
$$

$$
\ln B
=
2\pi\lambda_{из}R_н^{L}
\frac{t_п-t_в}{t_н-t_п}
$$

где $B$ — отношение наружного диаметра изоляции к внутреннему диаметру изоляции.

Требуемая толщина изоляции определяется по методике, изложенной в В.2.3.

В расчетах температуру наружной среды $t_н$ следует принимать равной температуре воздуха в помещении.

```python
import math
```

```python
def b_iz(d_n_pipe,
         B1, 
         ):
    return ((d_n_pipe * (B1 - 1)) / 2)

def ln_B1(lambda_iz,
         t_air, 
         t_surface, 
         t_inside,
         resistance_outside, 
         ):
    return (2 * math.pi * lambda_iz * resistance_outside * ((t_surface - t_inside)/(t_air - t_surface)))

def R_out_L(d_is_pipe, 
          lambda_out = 7
          ):
    return (1 / (math.pi * d_is_pipe * lambda_out))

 
```

```python
d_pipe = 28e-3
d_is = 19e-3
d_is_pipe = d_pipe + 2*d_is

t_inside_x = 5
t_surface_x = 25
t_air_x = 30

lambda_out = 7
lambda_iz = 0.034

r_out_L_x = R_out_L(d_is_pipe=d_is_pipe,  lambda_out=lambda_out)

ln_B1_x = ln_B1(
    lambda_iz=lambda_iz,
    t_surface=t_surface_x,
    t_inside=t_inside_x, 
    t_air=t_air_x, 
    resistance_outside=r_out_L_x
)

print(f"r_out_L_x = {r_out_L_x}" )
print(f"ln_B1_x = {ln_B1_x}" )
print(f"B_x = {math.exp(ln_B1_x)}" )

b_iz_x = b_iz(
    d_n_pipe=d_pipe,
    B1=math.exp(ln_B1_x)
)

print(f"b_iz_x = {b_iz_x:.4f}" )
```
